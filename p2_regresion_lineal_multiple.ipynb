```python
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Datos

X = np.array([
    [15, 10],
    [22, 5],
    [9, 9],
    [12, 4],
    [18, 7]
])

y = np.array([
    100,
    65,
    82,
    60,
    91
])

```

```python
# Crear el modelo

modelo = LinearRegression()

# Entrenar el modelo

modelo.fit(X, y)

# Mostrar Pesos

print("Intercepto:")
print(modelo.intercept_)

print("\nPesos:")
print(modelo.coef_)
```

```python
# Realizar predicciones

y_pred = modelo.predict(X)


print("\nPredicciones:")

for i, prediccion in enumerate(y_pred):
    print(f"Estudiante {chr(65+i)}: {prediccion:.4f}")

# Metricas

mse = mean_squared_error(y, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y, y_pred)


print("\nMétricas:")

print(f"MSE: {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²: {r2:.4f}")
```
