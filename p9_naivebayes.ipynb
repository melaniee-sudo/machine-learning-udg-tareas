```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.naive_bayes import MultinomialNB

#base de datos
data = {
    'Edad': [
        'Joven', 'Joven', 'Mediana', 'Mayor', 'Mayor',
        'Mayor', 'Mediana', 'Joven', 'Joven', 'Mayor',
        'Joven', 'Mediana', 'Mediana', 'Mayor'
    ],

    'Ingresos': [
        'Altos', 'Altos', 'Altos', 'Medios', 'Bajos',
        'Bajos', 'Bajos', 'Medios', 'Bajos', 'Medios',
        'Medios', 'Medios', 'Altos', 'Medios'
    ],

    'Estudiante': [
        'No', 'No', 'No', 'No', 'Si',
        'Si', 'Si', 'No', 'Si', 'Si',
        'Si', 'No', 'Si', 'No'
    ],

    'Credito': [
        'Regular', 'Excelente', 'Regular', 'Regular', 'Regular',
        'Excelente', 'Excelente', 'Regular', 'Regular', 'Regular',
        'Excelente', 'Excelente', 'Regular', 'Excelente'
    ],

    'Compra': [
        'No', 'No', 'Si', 'Si', 'Si',
        'No', 'Si', 'No', 'Si', 'Si',
        'Si', 'Si', 'Si', 'No'
    ]
}

df = pd.DataFrame(data)

print(df)

#convertir las variables categoricas de texto a numeros
encoder = OneHotEncoder(sparse_output=False)

X_encoded = encoder.fit_transform(
    df[['Edad', 'Ingresos', 'Estudiante', 'Credito']]
)

#variable que queremos predecir
y = df['Compra']

#crear y entrenar el modelo Naive Bayes
modelo = MultinomialNB()
modelo.fit(X_encoded, y)

#nuevo cliente: mayor, Ingresos Bajos, No es estudiante, Crédito Excelente
nuevo_cliente = pd.DataFrame(
    [['Mayor', 'Bajos', 'No', 'Excelente']],
    columns=['Edad', 'Ingresos', 'Estudiante', 'Credito']
)

#convertir el nuevo cliente usando el mismo encoder
nuevo_cliente_encoded = encoder.transform(nuevo_cliente)

#Prediccion
prediccion = modelo.predict(nuevo_cliente_encoded)

print(f"\nLa predicción para el nuevo cliente es: {prediccion[0]}")
```
