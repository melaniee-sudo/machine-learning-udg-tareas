```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

#cargar la base de datos Wine
wine = load_wine()

X = wine.data
y = wine.target

#70% entrenamiento y 30% prueba
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

#normalizar los datos
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# KNN con distancia Euclidiana

knn_euclidiana = KNeighborsClassifier(
    n_neighbors=3,
    metric="euclidean"
)

knn_euclidiana.fit(X_train, y_train)

pred_euclidiana = knn_euclidiana.predict(X_test)

accuracy_euclidiana = accuracy_score(
    y_test,
    pred_euclidiana
)

# KNN con distancia Manhattan

knn_manhattan = KNeighborsClassifier(
    n_neighbors=3,
    metric="manhattan"
)

knn_manhattan.fit(X_train, y_train)

pred_manhattan = knn_manhattan.predict(X_test)

accuracy_manhattan = accuracy_score(
    y_test,
    pred_manhattan
)

#resultados
print("Accuracy Euclidiana:", accuracy_euclidiana)
print("Accuracy Manhattan:", accuracy_manhattan)
```
