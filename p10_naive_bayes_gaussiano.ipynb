```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import GaussianNB


#bd
datos = {
    'Peso': [
        150, 160, 170, 145, 155,
        180, 190, 185, 200, 195,
        100, 105, 110, 95, 115
    ],

    'Diametro': [
        7.1, 7.3, 7.5, 7.0, 7.2,
        8.1, 8.3, 8.0, 8.5, 8.2,
        5.1, 5.3, 5.5, 5.0, 5.4
    ],

    'Acidez': [
        4.5, 4.4, 4.6, 4.5, 4.3,
        3.8, 3.7, 3.9, 3.6, 3.8,
        2.2, 2.3, 2.1, 2.4, 2.2
    ],

    'Fruta': [
        'Manzana', 'Manzana', 'Manzana', 'Manzana', 'Manzana',
        'Naranja', 'Naranja', 'Naranja', 'Naranja', 'Naranja',
        'Limón', 'Limón', 'Limón', 'Limón', 'Limón'
    ]
}


# crear dataFrame
df = pd.DataFrame(datos)

print("Base de datos:")
print(df)


#variables de entrada y salida
X = df[['Peso', 'Diametro', 'Acidez']]
y = df['Fruta']


#convertir la variable de texto a números
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

print("\nClases:")
print(encoder.classes_)

print("\nClases convertidas a números:")
print(y_encoded)


#crear y entrenar modelo
modelo = GaussianNB()
modelo.fit(X, y_encoded)

nueva_fruta = pd.DataFrame({
    'Peso': [175],
    'Diametro': [7.8],
    'Acidez': [4.0]
})

#realizar prediccion
prediccion = modelo.predict(nueva_fruta)

fruta_predicha = encoder.inverse_transform(prediccion)

print("\nFruta predicha:", fruta_predicha[0])


#calcular probabilidades
probabilidades = modelo.predict_proba(nueva_fruta)

print("\nProbabilidades:")

for fruta, probabilidad in zip(encoder.classes_, probabilidades[0]):
    print(f"{fruta}: {probabilidad:.6f}")
```
