```python
import numpy as np
import matplotlib.pyplot as plt

#Datos simulados
rng = np.random.default_rng(42)
x = np.arange(1, 16) #15 voltajes

#RPM
ruido = rng.integers(-35, 36, size= 15)
y= 180*x + 40 + ruido

n = len(x)
for i in range(n):
  print (x[i], y[i])
```

```python
#Tabla de sumatorias

x2 = x**2
xy = x*y

sum_x = np.sum(x)
sum_y = np.sum(y)
sum_x2 = np.sum(x2)
sum_xy = np.sum(xy)

print("Voltaje  RPM  x2  xy")
for i in range(n):
  print (x[i], "      ", y[i], " ", x2[i], " ", xy[i])

print("\nSumatorias")
print("Σx =", sum_x, "  Σy =", sum_y, "  Σx² =", sum_x2, "  Σxy =", sum_xy)
```

```python
#Regresion lineal
m = (n*sum_xy - sum_x*sum_y) / (n*sum_x2 - sum_x**2)
b = (sum_y - m*sum_x) / n

print("\nPendiente m =", m)
print("Interseccion b =", b)

print("\nEcuacion:")
print("RPM =", m, "* Voltaje +", b)

```

```python
#Prediccion para 15V
rpm_15 = m*n + b
#solo porque n tambien vale 15, so es lo mismo
print("\nRPM aproximadas a 15 V =", rpm_15)
```

```python
#Grafica
plt.scatter(x, y, label= "Datos simulados")
plt.plot(
    x,
    m*x +b,
    label= "Regresion lineal"
)

plt.xlabel("Voltaje (V)")
plt.ylabel("Velocidad (RPM)")
plt.title("Caracterizacion lineal de un motor DC")
plt.legend()
plt.grid()

plt.show()
```
